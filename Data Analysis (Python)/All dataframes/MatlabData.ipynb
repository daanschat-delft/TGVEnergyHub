```python
import pandas as pd
from scipy.io import savemat

# Load the pkl file
df = pd.read_pickle("dataframes-1-min-2/PV_Inverter_Active_Power.pkl")
# Convert 'time' to ISO 8601 strings

# Ensure 'time' column is a datetime object
df['time'] = pd.to_datetime(df['time'])

# Set 'time' column as the index
df.set_index('time', inplace=True)

# Resample to 1-second intervals
# Use interpolation to fill the missing values
df_resampled = df.resample('1S').interpolate(method='linear')

# If you need to reset the index back
df_resampled = df_resampled.reset_index()


#df['time'] = df['time'].astype(str)

# Save to .mat for MATLAB
savemat("PV_inverter.mat", {"time": df_resampled['time'].tolist(), "other_columns": df_resampled.to_dict(orient="list")})


#savemat("PV_inverter.mat", {"data": data.to_dict(orient="list")})
```
